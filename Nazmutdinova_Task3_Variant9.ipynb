Задача 1 (создание класса, конструктор, метод)

Текст задачи:
В ресторане ведётся учёт блюд. Создайте класс Dish с полями name (название), category (категория: суп, горячее, десерт) и price (цена). Реализуйте конструктор __init__, который принимает эти три значения и сохраняет их в объекте. Добавьте метод info(), который возвращает строку с информацией о блюде в формате: Название — категория, цена. Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: Борщ — суп, 350 ₽. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр self в методах.

Текст задачи:
У каждого заказа в ресторане есть внутренний номер, который нельзя менять или читать напрямую. Создайте класс Order, в котором номер заказа хранится в приватном поле __order_id. Реализуйте метод set_order_id(value), предназначенный только для установки номера при оформлении заказа, и метод get_order_id() для чтения значения. Повторная попытка изменить номер после оформления должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте заказ, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать order.__order_id — будет ошибка AttributeError, но через name mangling (order._Order__order_id) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

```python
class Order:
    def __init__(self):
        self.__order_id = None

    def set_order_id(self, value):
        if self.__order_id is None:
            self.__order_id = value

    def get_order_id(self):
        return self.__order_id

order = Order()
order.set_order_id(101)
print(order.get_order_id())

order.set_order_id(202)
print(order.get_order_id())

try:
    print(order.__order_id)
except AttributeError as e:
    print(f"Ошибка: {e}")

print(order._Order__order_id)
```

• Почему «игнорировать без ошибки» иногда лучше: Это защищает программу от аварийного завершения (падения) в ситуациях, когда действие не критично для логики приложения, а повторный вызов может быть случайным дублированием со стороны интерфейса.

• Почему приватность в Python — это договорённость: Механизм name mangling (искажение имён при добавлении __) создан для предотвращения случайного переопределения полей в наследниках, а не для защиты от взлома. Язык доверяет разработчикам: если поле начинается с подчёркивания, его не стоит трогать снаружи, хотя техническая возможность через специальный синтаксис остаётся.

Текст задачи:
Ресторан рассчитывает стоимость приготовления для разных типов блюд. Создайте классы Soup, MainCourse и Dessert, у каждого из которых есть метод cook_time(), возвращающий время приготовления в минутах: для супа — 30, для горячего — 45, для десерта — 15. Напишите функцию show_cook_time(dish), которая принимает любое блюдо и выводит время его приготовления. Убедитесь, что функция одинаково работает со всеми тремя типами объектов, не зная заранее, какой именно объект ей передан. Объясните, как Python определяет, какой именно метод вызвать, и что такое «утиная типизация» — детально она разбирается в задаче 30.

```python
class Soup:
    def cook_time(self):
        return 30

class MainCourse:
    def cook_time(self):
        return 45

class Dessert:
    def cook_time(self):
        return 15

def show_cook_time(dish):
    print(f"Время приготовления: {dish.cook_time()} минут")

soup = Soup()
main = MainCourse()
dessert = Dessert()

show_cook_time(soup)
show_cook_time(main)
show_cook_time(dessert)
```

• Как Python определяет метод: Так как в Python типизация динамическая, в момент вызова dish.cook_time() интерпретатор заглядывает внутрь переданного объекта, находит у него этот метод и выполняет его.

• Что такое «утиная типизация»: Это принцип, согласно которому тип объекта определяется не его положением в иерархии классов, а его поведением (интерфейсом). Если объект «ходит как утка и крякает как утка» (в данном случае — имеет нужный метод cook_time()), Python считает его уткой (подходящим блюдом), даже если эти классы никак не связаны общим родителем.

Продемонстрируйте обращение к каждому из полей извне класса: к name — свободно, к _hall — можно, но не рекомендуется, к __salary — вызовет ошибку. Дополнительно покажите доступ к приватному полю через name mangling (_RestaurantManager__salary) — так вы увидите, что «приватность» в Python условна. Текст задачи:
Ресторану нужна утилита для проверки номеров столиков, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс RestaurantUtils со статическим методом is_valid_table(number), который возвращает True, если число находится в диапазоне от 1 до 50, и False — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: RestaurantUtils.is_valid_table(12). Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

```python
class RestaurantUtils:
    @staticmethod
    def is_valid_table(number):
        return 1 <= number <= 50

print(RestaurantUtils.is_valid_table(12))
print(RestaurantUtils.is_valid_table(60))
```

• Отличие от обычного метода: Обычный метод жестко привязан к конкретному экземпляру и первым аргументом всегда принимает self.

• Отличие от метода класса: Метод класса (декоратор @classmethod) привязан к самому классу и первым аргументом принимает ссылку на класс cls. Статический метод не знает ничего ни об объекте, ни о классе — это просто изолированная функция, помещённая внутрь класса для удобства группировки.

• Когда использовать: Его стоит использовать для вспомогательных, утилитарных функций (валидаторы данных, математические конвертеры), логика которых независима от состояния объектов класса.

Задача 8 (композиция)

Текст задачи:
У каждого блюда есть тарелка для подачи — блюдо не может быть подано без тарелки, и тарелка создаётся вместе с блюдом. Создайте класс Plate с методом show(), который выводит строку «Круглая тарелка». Создайте класс Dish, который содержит объект класса Plate внутри себя как поле (композиция) — объект Plate создаётся прямо в конструкторе блюда. Метод show_dish() у блюда должен выводить название блюда и вызывать метод show() у его тарелки, демонстрируя, что блюдо «владеет» своей тарелкой. Объясните, почему это именно композиция, а не агрегация (разница разбирается в задаче 18).

```python
class Plate:
    def show(self):
        return "Круглая тарелка"

class Dish:
    def __init__(self, name):
        self.name = name
        self.plate = Plate()

    def show_dish(self):
        print(f"Блюдо: {self.name}, Подача: {self.plate.show()}")

dish = Dish("Стейк")
dish.show_dish()
```

• Почему это композиция, а не агрегация: В композиции жизненный цикл вложенного объекта жестко управляется контейнером. Объект Plate создаётся внутри конструктора Dish. Если уничтожить объект Dish, его внутренняя тарелка Plate также будет автоматически уничтожена. Они не могут существовать раздельно в логике этой связи. При агрегации объект тарелки создавался бы снаружи и передавался бы в готовом виде в конструктор.



Задача 9 (переопределение методов)

Текст задачи:
В ресторане есть базовый класс MenuItem с методом get_info(), который возвращает строку «Позиция меню». Унаследуйте от него класс Dish и переопределите метод get_info() так, чтобы он возвращал строку «Блюдо: <название>». Создайте объект класса Dish, вызовите метод get_info() и убедитесь, что используется версия из дочернего класса, а не из родительского. Объясните разницу между переопределением (override) и перегрузкой (overload) — и почему в Python «настоящей» перегрузки методов нет.

```python
class MenuItem:
    def get_info(self):
        return "Позиция меню"

class Dish(MenuItem):
    def __init__(self, name):
        self.name = name

    def get_info(self):
        return f"Блюдо: {self.name}"

dish = Dish("Тирамису")
print(dish.get_info())
```

• Разница между переопределением и перегрузкой: Переопределение (override) — это изменение логики родительского метода в дочернем классе с сохранением имени и сигнатуры. Перегрузка (overload) — это создание нескольких функций с одним именем, но с разным набором/типом принимаемых аргументов в рамках одного класса.

• Почему в Python нет настоящей перегрузки: Переменные в Python не имеют статических типов, а функции являются объектами первого класса. Если в одном классе объявить два метода с одинаковым именем, последний просто перезапишет первый в словаре атрибутов класса (__dict__).


Задача 10 (имитация перегрузки)

Текст задачи:
Официанту нужен метод для оформления заказов, который мог бы работать в разных ситуациях: иногда известно только блюдо, иногда блюдо и количество порций, иногда ещё и столик. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и *args. Создайте класс OrderDesk с методом take(dish, quantity=None, table=None), который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Оформлен заказ на блюдо: <название>», «Оформлен заказ: <название> × <количество>» или «Оформлен заказ: <название> × <количество> на столик <номер>». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

```python
class OrderDesk:
    def take(self, dish, quantity=None, table=None):
        if quantity is not None and table is not None:
            print(f"Оформлен заказ: {dish} × {quantity} на столик {table}")
        elif quantity is not None:
            print(f"Оформлен заказ: {dish} × {quantity}")
        else:
            print(f"Оформлен заказ на блюдо: {dish}")

desk = OrderDesk()
desk.take("Пицца")
desk.take("Пицца", 2)
desk.take("Пицца", 2, 5)
```

Почему в Python выбрали такой подход: Благодаря наличию динамической типизации, аргументов по умолчанию, а также поддержке позиционных (*args) и именованных (**kwargs) аргументов, один-единственный метод может гибко обрабатывать любые комбинации входных данных. Это делает код более лаконичным и избавляет от необходимости дублировать имена методов для разных типов входных параметров.

Задача 11 (магические методы __str__ и __repr__)

Текст задачи:
Блюда часто выводят на экран — в меню, заказах, отчётах. По умолчанию print(dish) выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс Dish магический метод __str__, который возвращает строку вида «Блюдо "Борщ" (суп, 350 ₽)». Убедитесь, что при вызове print(dish) выводится именно эта строка. Дополнительно реализуйте метод __repr__, который возвращает техническое представление объекта, например Dish('Борщ', 'суп', 350) — оно используется при выводе списка блюд и в отладчике. Поясните разницу между __str__ и __repr__ и когда какой используется: __str__ — для человека, __repr__ — для разработчика.

```python
class Dish:
    def __init__(self, name, category, price):
        self.name = name
        self.category = category
        self.price = price

    def __str__(self):
        return f'Блюдо "{self.name}" ({self.category}, {self.price} ₽)'

    def __repr__(self):
        return f"Dish('{self.name}', '{self.category}', {self.price})"

dish = Dish("Борщ", "суп", 350)
print(dish)

dishes_list = [dish]
print(dishes_list)
```

• Разница между __str__ и __repr__: Метод __str__ предназначен для создания красивого, удобного и понятного текстового представления объекта, которое будет приятно читать конечному пользователю приложения. Метод __repr__ формирует однозначное, техническое описание объекта, по возможности копирующее синтаксис его создания на Python, чтобы облегчить процесс поиска ошибок.

• Когда какой используется: __str__ автоматически вызывается встроенными функциями print(), str() и при форматировании строк для людей.

• __repr__ применяется по умолчанию в интерактивной консоли, отладчиках, а также при отображении коллекций (например, когда объект находится внутри списка или словаря).

Текст задачи:
В ресторане может оказаться несколько записей об одном и том же блюде — например, при повторном добавлении в меню. Нужно, чтобы программа понимала, что это «одно и то же блюдо», и не дублировала его в меню. Реализуйте в классе Dish магический метод __eq__, чтобы два блюда считались равными, если совпадают их название и категория. Дополнительно реализуйте __hash__, чтобы блюда можно было хранить во множестве set и использовать как ключи в словаре. Проверьте, что два одинаковых блюда попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении __eq__ нужно также переопределять __hash__ — что произойдёт, если этого не сделать.

```python
class Dish:
    def __init__(self, name, category, price):
        self.name = name
        self.category = category
        self.price = price

    def __eq__(self, other):
        if not isinstance(other, Dish):
            return NotImplemented
        return self.name == other.name and self.category == other.category

    def __hash__(self):
        return hash((self.name, self.category))

dish1 = Dish("Борщ", "суп", 350)
dish2 = Dish("Борщ", "суп", 400)
dish3 = Dish("Цезарь", "салат", 450)

menu_set = {dish1, dish2, dish3}
print(len(menu_set))
```

• Почему при переопределении __eq__ нужно переопределять __hash__: Поведение хэш-таблиц (словари dict и множества set) опирается на контракт: если два объекта равны по мнению __eq__, они обязаны возвращать абсолютно одинаковые хэш-значения.

• Что произойдёт, если этого не сделать: Если переопределить только __eq__, Python автоматически установит __hash__ = None, сделав объекты этого класса «нехэшируемыми». Попытка положить такое блюдо в set или использовать как ключ в dict вызовет критическую ошибку TypeError.


Задача 13 (магические методы __add__ и __len__)

Текст задачи:
В ресторане есть меню, и иногда нужно объединять их (например, при объединении двух ресторанов) или узнавать, сколько блюд в меню. Создайте класс Menu (меню), в котором хранится список блюд. Реализуйте магический метод __add__ для объединения двух меню в одно (возвращается новое меню со всеми блюдами) и метод __len__, возвращающий количество блюд. Продемонстрируйте работу: создайте два меню, объедините их через menu3 = menu1 + menu2 и выведите len(menu3). Объясните, почему __add__ возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

```python
class Menu:
    def __init__(self, dishes=None):
        self.dishes = dishes if dishes is not None else []

    def __add__(self, other):
        if not isinstance(other, Menu):
            return NotImplemented
        return Menu(self.dishes + other.dishes)

    def __len__(self):
        return len(self.dishes)

menu1 = Menu(["Борщ", "Солянка"])
menu2 = Menu(["Тирамису", "Чизкейк"])

menu3 = menu1 + menu2
print(len(menu3))
```

• Почему __add__ возвращает новый объект: Оператор сложения + по своей математической природе должен быть неизменяемым (immutable) для исходных операндов. Когда мы складываем 5 + 3, пятерка не превращается в восьмерку, а рождается новое число. Если бы мы хотели изменить существующий объект, следовало бы использовать метод __iadd__ для оператора +=.

• Как магические методы делают объекты «похожими» на встроенные типы: Магические методы позволяют кастомным классам перегружать стандартное поведение языка. Реализуя такие методы, разработчик дает возможность применять к своим объектам привычные встроенные функции (len()), операторы (+, ==) и циклы (for), делая интерфейс классов интуитивно понятным.



Задача 14 (свойство @property)

Текст задачи:
В ресторане важно, чтобы цена блюда была корректной — нельзя допустить отрицательную цену или ноль. Перепишите класс Dish так, чтобы поле price было реализовано через свойство @property. Проверки: цена должна быть числом (int или float), и она должна быть строго больше нуля. При нарушении должно выбрасываться исключение ValueError с понятным сообщением. Покажите, что обращение к dish.price работает как к обычному атрибуту (без скобок), но при этом выполняется проверка: dish.price = -100 → ошибка, dish.price = 0 → ошибка, dish.price = "abc" → ошибка, dish.price = 350 → успех. Объясните, чем @property удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

```python
class Dish:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if not isinstance(value, (int, float)):
            raise ValueError("Цена должна быть числом")
        if value <= 0:
            raise ValueError("Цена должна быть строго больше нуля")
        self._price = value

dish = Dish("Борщ", 350)
print(dish.price)

for invalid_price in [-100, 0, "abc"]:
    try:
        dish.price = invalid_price
    except ValueError as e:
        print(f"Ошибка для {invalid_price}: {e}")

dish.price = 400
print(dish.price)
```

• Чем @property удобнее классических геттеров и сеттеров: Декоратор @property позволяет сохранить чистый и элегантный синтаксис доступа к полям через точку (obj.attr = value) без громоздких вызовов методов в стиле obj.set_price(value). Это дает возможность внедрять валидацию данных постфактум, не ломая обратную совместимость стороннего кода, который использовал обычные публичные атрибуты.

• Что такое «вычисляемый атрибут»: Это свойство класса, значение которого не хранится в памяти постоянно, а вычисляется динамически «на лету» в момент обращения к нему (например, свойство full_name, которое склеивает имя и фамилию).


Задача 15 (классовый метод @classmethod)

Текст задачи:
В ресторан часто поступают данные о блюдах в виде строк — например, из файла или базы данных: "Борщ;суп;350". Добавьте в класс Dish классовый метод from_string(cls, data), который создаёт объект блюда из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание блюда через этот метод и вывод результата. Объясните разницу между @classmethod, @staticmethod и обычным методом: обычный метод получает первым аргументом экземпляр (self) и работает с конкретным объектом, @classmethod получает сам класс (cls) и часто используется как альтернативный конструктор, @staticmethod не получает ни self, ни cls и представляет собой просто утилиту. Расскажите, когда какой использовать.

```python
class Dish:
    def __init__(self, name, category, price):
        self.name = name
        self.category = category
        self.price = price

    @classmethod
    def from_string(cls, data):
        name, category, price_str = data.split(";")
        return cls(name, category, float(price_str))

dish = Dish.from_string("Борщ;суп;350")
print(dish.name, dish.category, dish.price)
```

• Разница между методами: Обычный метод привязан к объекту и принимает self. Используется, когда нужно читать или изменять состояние конкретного экземпляра.

• Классовый метод (@classmethod) привязан к самому классу, принимает cls. Используется для создания альтернативных конструкторов или для работы с данными, глобальными для всего класса.

• Статический метод (@staticmethod) не привязан ни к объекту, ни к классу, не принимает неявных аргументов. Используется как изолированная вспомогательная функция, которая относится к предметной области класса, но не нуждается в его внутренней структуре.

• Когда какой использовать: Обычный — для логики конкретного объекта. Классовый — для фабричных методов и альтернативной инициализации. Статический — для чистых функций-утилит (валидация, парсинг), сгруппированных внутри класса для порядка.

Создайте собственное исключение DishNotFoundError(Exception) и используйте его в классе Restaurant (ресторан). Метод add_to_order(order, dish_name) должен добавлять блюдо в заказ, если оно есть в меню, и выбрасывать DishNotFoundError с понятным сообщением (например, с указанием названия), если блюда нет. Продемонстрируйте обработку исключения через try/except и объясните, зачем создавать собственные исключения вместо использования стандартных — как это помогает ловить именно свои ошибки и отделять их от чужих.

```python
class DishNotFoundError(Exception):
    def __init__(self, dish_name):
        self.dish_name = dish_name
        super().__init__(f"Блюдо '{dish_name}' не найдено в меню ресторана.")


class Restaurant:
    def __init__(self, menu):
        self.menu = menu

    def add_to_order(self, order, dish_name):
        if dish_name not in self.menu:
            raise DishNotFoundError(dish_name)
        order.append(dish_name)


restaurant = Restaurant(["Борщ", "Салат"])
my_order = []

try:
    restaurant.add_to_order(my_order, "Паста")
except DishNotFoundError as e:
    print(f"Перехвачено исключение: {e}")
    print(f"Проблема возникла из-за блюда: {e.dish_name}")
```

• Зачем создавать собственные исключения: Кастомные исключения делают архитектуру приложения выразительной, сообщая точную бизнес-причину сбоя (ошибка логики ресторана, а не системный сбой работы с памятью).

• Как это помогает в отлове ошибок: Это позволяет перехватывать в блоках except строго определенные ошибки бизнес-логики, не затрагивая непредвиденные системные баги.

• Если бы мы использовали встроенный ValueError, то могли бы случайно перехватить ошибку от неправильного преобразования типов где-то глубоко в коде, скрыв реальный баг.

Текст задачи:
В ресторане работает человек, который одновременно и официант, и повар — например, универсальный сотрудник, который и готовит, и обслуживает. Продемонстрируйте порядок разрешения методов через ChefWaiter.__mro__ и объясните, какой метод get_role() будет вызван и почему. Расскажите, что такое MRO (Method Resolution Order), как Python ищет методы при множественном наследовании и что такое «проблема ромба» (diamond problem).

```python
class Chef:
    def get_role(self):
        return "Повар"

class Waiter:
    def get_role(self):
        return "Официант"

class ChefWaiter(Chef, Waiter):
    pass

employee = ChefWaiter()
print(employee.get_role())
print(ChefWaiter.__mro__)
```

• Какой метод get_role() будет вызван и почему: Будет вызван метод класса Chef (вернет "Повар"). Это происходит потому, что в объявлении класса ChefWaiter(Chef, Waiter) класс Chef указан первым слева направо, и в порядке MRO он стоит ближе к дочернему классу.

• Что такое MRO (Method Resolution Order): Это строго определенный линейный порядок (список), в котором Python просматривает базовые классы при поиске нужного атрибута или метода. Для вычисления MRO в Python используется современный алгоритм C3-линеаризации, гарантирующий предсказуемость и соблюдение монотонности и локального порядка приоритетов.

• Что такое «проблема ромба» (diamond problem): Это классическая неопределенность в языках с множественным наследованием, когда один дочерний класс наследуется от двух родителей, которые, в свою очередь, унаследованы от одного общего предка. Если метод определен у предка и переопределен только у одного из родителей, возникает путаница, чью версию метода должен запустить дочерний класс. Python решает эту проблему однозначно с помощью цепочки MRO.



Задача 20 (@dataclass и миксин)

Текст задачи:
Класс Dish содержит только данные и простые методы — его удобно переписать через декоратор @dataclass, чтобы не писать вручную конструктор, __repr__ и __eq__. Перепишите класс Dish через @dataclass и сравните с версией из задачи 1. Дополнительно создайте миксин LoggableMixin с методом log(), который выводит имя класса и текущее время (используйте from datetime import datetime). Подключите миксин к классу Dish через множественное наследование, причём миксин должен идти первым в списке родителей: class Dish(LoggableMixin): .... Продемонстрируйте работу миксина. Объясните, что такое миксин и чем он отличается от обычного родительского класса: миксин — это «примесь» с одной узкой функциональностью, он не предназначен для самостоятельного использования и обычно не имеет собственного состояния.

```python
from dataclasses import dataclass
from datetime import datetime

class LoggableMixin:
    def log(self):
        print(f"[{datetime.now()}] Вызван лог для класса: {self.__class__.__name__}")

@dataclass
class Dish(LoggableMixin):
    name: str
    category: str
    price: float

dish = Dish("Борщ", "суп", 350.0)
print(dish)
dish.log()
```

• Сравнение с версией из задачи 1: Использование @dataclass избавляет разработчика от необходимости писать шаблонный код (boilerplate). Он автоматически генерирует методы __init__, __repr__ и __eq__ на основе объявленных аннотаций типов полей.

• Что такое миксин (примесь) и его отличия от обычного класса: Миксин — это специфический класс, предназначенный исключительно для добавления определенного изолированного поведения (методов) в другие классы через множественное наследование. В отличие от полноценного родительского класса, миксин решает только одну узкую задачу, не имеет логического смысла как самостоятельная сущность (его экземпляры никогда не создаются напрямую) и, как правило, не хранит собственного внутреннего состояния (у него нет своего конструктора __init__).

При таком объёме важна экономия памяти: каждый объект Dish по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс Dish атрибут __slots__ = ("name", "category", "price"). Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, dish.weight = 250 вызовет AttributeError). Сравните размер объекта с __slots__ и без него через sys.getsizeof — но учтите, что getsizeof показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте pympler.asizeof или оцените по количеству полей. Объясните, как __slots__ экономит память, что теряется при его использовании (наследование, __dict__, @property) и в каких случаях его стоит применять.

```python
import sys

class DishWithSlots:
    __slots__ = ("name", "category", "price")
    def __init__(self, name, category, price):
        self.name = name
        self.category = category
        self.price = price

class DishWithoutSlots:
    def __init__(self, name, category, price):
        self.name = name
        self.category = category
        self.price = price

dish_slots = DishWithSlots("Борщ", "суп", 350)
dish_no_slots = DishWithoutSlots("Борщ", "суп", 350)

print(f"Размер оболочки со slots: {sys.getsizeof(dish_slots)} байт")
print(f"Размер оболочки без slots: {sys.getsizeof(dish_no_slots)} байт")

try:
    dish_slots.weight = 250
except AttributeError as e:
    print(f"Ошибка добавления поля: {e}")
```

• Как __slots__ экономит память: По умолчанию Python динамически создает словарь __dict__ для каждого объекта, чтобы хранить его атрибуты. Словари в Python оптимизированы под скорость доступа, из-за чего они резервируют избыточную память. Атрибут __slots__ заменяет этот динамический словарь на фиксированный, оптимизированный на уровне C массив ссылок строго заданного размера.

• Что теряется при использовании __slots__: У объектов полностью пропадает встроенный словарь __dict__, из-за чего блокируется возможность динамически добавлять новые, не описанные заранее атрибуты.

• Ограничивается и усложняется наследование: дочерние классы автоматически получат __dict__, если они заново не объявят свой собственный __slots__. При этом стандартные свойства @property работают нормально, так как они опираются на дескрипторы, а не на __dict__.

• В каких случаях стоит применять: Его имеет смысл внедрять исключительно в ситуациях, когда в оперативной памяти приложения одновременно создаются и удерживаются сотни тысяч или миллионы экземпляров легковесных классов, содержащих строго фиксированный набор данных.


Задача 22 (дескрипторы)

Текст задачи:
Свойство @property — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор RangeDescriptor(min_value, max_value), который хранит значение в instance.__dict__ (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая ValueError при нарушении. Используйте этот дескриптор в классе Dish для полей price (например, диапазон 1–100000) и weight (диапазон 1–500), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» @property, зачем нужен метод __set_name__ и в чём преимущество дескриптора перед @property, если одинаковая проверка повторяется в десяти полях

```python
class RangeDescriptor:
    def __init__(self, min_value, max_value):
        self.min_value = min_value
        self.max_value = max_value

    def __set_name__(self, owner, name):
        self.name = name

    def __get__(self, instance, owner):
        if instance is None:
            return self
        return instance.__dict__.get(self.name)

    def __set__(self, instance, value):
        if not isinstance(value, (int, float)):
            raise ValueError(f"{self.name} должно быть числом")
        if not (self.min_value <= value <= self.max_value):
            raise ValueError(f"{self.name} должно быть в диапазоне от {self.min_value} до {self.max_value}")
        instance.__dict__[self.name] = value

class Dish:
    price = RangeDescriptor(1, 100000)
    weight = RangeDescriptor(1, 5000)

    def __init__(self, name, price, weight):
        self.name = name
        self.price = price
        self.weight = weight

dish = Dish("Стейк", 1500, 350)
print(f"Цена: {dish.price}, Вес: {dish.weight}")

try:
    dish.price = 200000
except ValueError as e:
    print(f"Ошибка валидации цены: {e}")

try:
    dish.weight = -10
except ValueError as e:
    print(f"Ошибка валидации веса: {e}")
```

• Как дескрипторы работают под капотом @property: Декоратор @property — это встроенный синтаксический сахар, который под капотом создает готовый объект-дескриптор данных. Когда мы пишем obj.prop, срабатывает метод дескриптора __get__, а при присваивании obj.prop = val вызывается метод __set__.

• Зачем нужен метод __set_name__: Этот метод автоматически вызывается в момент создания содержащего класса. Он передает дескриптору имя переменной (например, "price" или "weight"), под которым дескриптор зарегистрирован в классе. Это избавляет от необходимости вручную прописывать текстовое имя поля в конструкторе дескриптора и позволяет использовать его как ключ для хранения данных внутри instance.__dict__.

• В чем преимущество дескриптора перед @property: Дескриптор реализует важнейший принцип программирования DRY (Don't Repeat Yourself). Если одинаковая валидация (например, проверка числового диапазона) нужна для 10 полей, использование @property заставит написать 10 одинаковых пар геттеров и сеттеров, раздувая код. Дескриптор же пишется один раз, а затем переиспользуется в одну строчку для любого числа полей.


Задача 23 (метаклассы)

Текст задачи:
В ресторане должен быть только один реестр заказов — если создать второй, данные могут разойтись. Создайте метакласс SingletonMeta, который гарантирует, что у класса Restaurant будет только один экземпляр. Проверьте, что Restaurant() дважды возвращает один и тот же объект (сравните через is). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

```python
class SingletonMeta(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class Restaurant(metaclass=SingletonMeta):
    def __init__(self, name="Главный Ресторан"):
        self.name = name

r1 = Restaurant()
r2 = Restaurant()

print(r1 is r2)
```

• Что такое метакласс и его связь с классом: В Python всё является объектами, включая сами классы. Метакласс — это «класс для классов», то есть фабрика, которая конструирует и определяет поведение обычных классов в момент их объявления. Точно так же, как обычный класс порождает свои экземпляры (объекты), метакласс порождает свои экземпляры, которыми и являются сами классы. По умолчанию все классы в Python имеют метакласс type.

• Зачем нужны метаклассы: Метаклассы дают возможность глобально вмешиваться в процесс создания и конфигурации классов. С их помощью можно автоматически регистрировать создаваемые классы в реестре, проверять наличие обязательных методов (продвинутая валидация API), автоматически модифицировать или добавлять новые атрибуты «на лету» в момент импорта модулей. Это незаменимо при создании сложных ORM-систем (как в Django) или систем управления зависимостями в фреймворках.


Задача 24 (паттерн Singleton)

Текст задачи:
Реализуйте класс OrderRegistry через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода __new__ (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; __new__ проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

```python
class SingletonMeta(type):
    _instances = {}
    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class OrderRegistryMeta(metaclass=SingletonMeta):
    pass

class OrderRegistryNew:
    _instance = None
    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

reg_meta1 = OrderRegistryMeta()
reg_meta2 = OrderRegistryMeta()
print(f"Метакласс: {reg_meta1 is reg_meta2}")

reg_new1 = OrderRegistryNew()
reg_new2 = OrderRegistryNew()
print(f"Через __new__: {reg_new1 is reg_new2}")
```

• В каких случаях паттерн Singleton полезен: Паттерн незаменим для глобальных координаторов системы, которые должны строго в единственном числе управлять разделяемыми ресурсами. Типичные примеры: единая точка доступа к конфигурационным файлам приложения, централизованный логгер, записывающий события в один файл, или кэш данных в оперативной памяти, общий для всех модулей программы.

• В каких случаях паттерн Singleton вреден: Одиночки превращаются в глобальные переменные, что нарушает чистоту архитектуры. Они сильно усложняют юнит-тестирование, так как сохраняют скрытое глобальное состояние между тестами, делая их взаимозависимыми. Они маскируют реальные зависимости внутри функций (компоненты запрашивают одиночку напрямую вместо явного получения через аргументы) и жестко связывают код, мешая при необходимости быстро заменить реализацию объекта на заглушку (mock-объект) или расширить систему до нескольких параллельных реестров.
