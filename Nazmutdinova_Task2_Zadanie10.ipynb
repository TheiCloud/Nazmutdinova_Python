КЛАССЫ И ОБЪЕКТЫ. Задача 10: Создайте класс Course с полями название курса, код курса и количество кредитов. Реализуйте методы update_credits(new_credits), который обновляет количество кредитов, и display_info(), который выводит информацию о курсе.


```python
class Course:
    def __init__(self, course_name, course_code, credits):
        self.course_name = course_name
        self.course_code = course_code
        self.credits = credits

    def update_credits(self, new_credits):
        self.credits = new_credits

    def display_info(self):
        print(f"Курс: {self.course_name}")
        print(f"Код курса: {self.course_code}")
        print(f"Количество кредитов: {self.credits}")

course1 = Course("Программирование на Python", "CS-101", 4)
course1.display_info()

print("\n--- Обновление кредитов ---")
course1.update_credits(5)
course1.display_info()
```

НАСЛЕДОВАНИЕ. Задача 10. Определите класс Course, который наследуется от класса Event. Класс Course должен добавлять поля код курса и количество кредитов, а также методы для обновления количества кредитов.

```python
class Event:
    def __init__(self, name):
        self.name = name

class Course(Event):
    def __init__(self, name, course_code, credits):
        super().__init__(name)
        self.course_code = course_code
        self.credits = credits

    def update_credits(self, new_credits):
        self.credits = new_credits

course2 = Course("Высшая математика", "MATH-201", 3)
print(f"Название (из Event): {course2.name}")
print(f"Код: {course2.course_code}, Кредиты: {course2.credits}")
```

ИНКАПСУЛЯЦИЯ. Задача 10. Создать класс Course, в котором поля courseName, courseCode и credits будут приватными. Реализовать методы для доступа и изменения количества кредитов с проверкой на допустимые значения.


```python
class Course:
    def __init__(self, course_name, course_code, credits):
        self.__courseName = course_name
        self.__courseCode = course_code
        self.__credits = credits if credits >= 0 else 0

    def get_credits(self):
        return self.__credits

    def set_credits(self, new_credits):
        if isinstance(new_credits, (int, float)) and new_credits >= 0:
            self.__credits = new_credits
            print(f"Количество кредитов успешно изменено на: {new_credits}")
        else:
            print("Ошибка: Количество кредитов должно быть неотрицательным числом!")

course3 = Course("Алгоритмы и структуры данных", "CS-202", 6)

course3.set_credits(-3)

course3.set_credits(5)
print(f"Текущие кредиты: {course3.get_credits()}")
```
