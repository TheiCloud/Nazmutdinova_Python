3. Дана длина ребра куба. Найти площадь грани, площадь полной поверхности и объем этого куба

```python
# Ввод длины ребра куба
a = float(input("Введите длину ребра куба: "))

if a <= 0:
    print("Ошибка: длина ребра должна быть положительным числом!")
else:
    # Вычисления
    s_face = a ** 2          # Площадь одной грани
    s_total = 6 * s_face     # Площадь полной поверхности
    v = a ** 3               # Объем куба

    # Вывод результатов
    print(f"Площадь грани: {s_face}")
    print(f"Площадь полной поверхности: {s_total}")
    print(f"Объем куба: {v}")
```

Определить, равен ли квадрат заданного трехзначного числа кубу суммы цифр этого числа.

```python
# Ввод трехзначного числа
num = int(input("Введите трехзначное число: "))

if not (100 <= abs(num) <= 999):
    print("Ошибка: введено не трехзначное число!")
else:
    # Находим цифры числа
    n = abs(num)
    digit1 = n // 100
    digit2 = (n // 10) % 10
    digit3 = n % 10

    # Считаем квадрат числа и куб суммы его цифр
    square_of_num = num ** 2
    cube_of_sum = (digit1 + digit2 + digit3) ** 3

    # Проверяем равенство
    result = (square_of_num == cube_of_sum)

    print(f"Квадрат числа ({square_of_num}) равен кубу суммы цифр ({cube_of_sum}): {result}")
```

Дан радиус круга \(R\). Определить, поместится ли правильный треугольник со стороной \(a\) в этом круге.

```python
import math

# Ввод радиуса круга и стороны треугольника
R = float(input("Введите радиус круга R: "))
a = float(input("Введите сторону правильного треугольника a: "))

# Радиус описанной около правильного треугольника окружности равен R_min = a / sqrt(3)
# Треугольник поместится в круг, если радиус круга R больше или равен R_min
r_needed = a / math.sqrt(3)

if R >= r_needed:
    print("Да, правильный треугольник поместится в этом круге.")
else:
    print("Нет, правильный треугольник не поместится в этом круге.")
```

Известны год и номер месяца рождения человека, а также год и номер месяца сегодняшнего дня (январь — 1 и т. д.). Определить возраст человека (число полных лет). В случае совпадения указанных номеров месяцев читать, что прошел полный год.

```python
# Ввод данных
birth_year = int(input("Введите год рождения: "))
birth_month = int(input("Введите номер месяца рождения (1-12): "))

current_year = int(input("Введите текущий год: "))
current_month = int(input("Введите текущий номер месяца (1-12): "))

# Базовый расчет разницы лет
age = current_year - birth_year

# Если текущий месяц меньше месяца рождения, то полный год еще не исполнился
if current_month < birth_month:
    age -= 1

print(f"Полный возраст человека: {age} лет/года")
```

```python
# Функция для проверки, является ли число простым
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

all_prime = True

print("Результаты вычислений по формуле x^2 + x + 41:")
# Цикл по x от 0 до 40 включительно
for x in range(41):
    result = x**2 + x + 41
    prime_status = is_prime(result)
    print(f"x = {x:2d} -> {result:4d} (Простое: {prime_status})")

    if not prime_status:
        all_prime = False

print("-" * 40)
print(f"Утверждение, что все числа простые: {all_prime}")
```

```python
import math

epsilon = 10**-3
S = 0.0
n = 1

while True:
    # Вычисление общего члена ряда a_n
    # Для больших n эффективнее вычислять через рекуррентное соотношение,
    # но для базовой точности 10^-3 можно посчитать "в лоб":
    a_n = (3**n * math.factorial(n)) / math.factorial(2 * n)

    # Если член ряда меньше заданной точности, прекращаем вычисления
    if a_n < epsilon:
        break

    S += a_n
    n += 1

print(f"Сумма ряда с точностью {epsilon} равна: {S}")
print(f"Количество учтенных членов ряда: {n - 1}")
```

```python
import math
from fractions import Fraction

def calculate_harmonic_sum(n):
    # Используем встроенный модуль fractions для точной работы с дробями
    total_sum = Fraction(0, 1)

    for i in range(1, n + 1):
        total_sum += Fraction(1, i)

    # Возвращаем числитель p и знаменатель q (они автоматически несократимы)
    return total_sum.numerator, total_sum.denominator

# Пример работы программы:
n = int(input("Введите натуральное число n: "))
if n <= 0:
    print("Число должно быть натуральным (больше 0)!")
else:
    p, q = calculate_harmonic_sum(n)
    print(f"Сумма ряда в виде несократимой дроби p/q: {p}/{q}")
```

Составить функцию, проверяющую, являются ли данные три числа взаимно простыми

```python
import math

def are_coprime_three(a, b, c):
    # Три числа взаимно просты, если их наибольший общий делитель (НОД) равен 1
    # math.gcd может принимать несколько аргументов в современных версиях Python
    return math.gcd(a, b, c) == 1

# Пример работы программы:
num1 = int(input("Введите первое число: "))
num2 = int(input("Введите второе число: "))
num3 = int(input("Введите третье число: "))

result = are_coprime_three(num1, num2, num3)
print(f"Числа являются взаимно простыми: {result}")
```

```python
def to_base_p(n, p):
    # Базовый случай рекурсии
    if n < p:
        return str(n)
    # Рекурсивный шаг: делим число на основание p и добавляем остаток в конец
    return to_base_p(n // p, p) + str(n % p)

# Пример работы программы:
number = int(input("Введите натуральное число: "))
base = int(input("Введите основание системы счисления p (от 2 до 9): "))

if not (2 <= base <= 9) or number < 0:
    print("Неверные входные данные!")
else:
    result_str = to_base_p(number, base)
    print(f"Число {number} в {base}-ичной системе счисления: {result_str}")
```
